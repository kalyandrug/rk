```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

# --- 1. Load and prep data ---
#df = pd.read_csv('housing_simple.csv')

binary_cols = ['mainroad', 'guestroom', 'basement', 'hotwaterheating', 'airconditioning', 'prefarea']
for col in binary_cols:
    df[col] = df[col].map({'yes': 1, 'no': 0})

df_encoded = pd.get_dummies(df, columns=['furnishingstatus'], drop_first=True)

X = df_encoded.drop('price', axis=1)
y = df_encoded['price']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)

# --- 2. Train model ---
model = LinearRegression()
model.fit(X_train, y_train)
preds = model.predict(X_test)

# --- 3. Plot 1: Actual vs Predicted scatter ---
plt.figure(figsize=(7, 6))
plt.scatter(y_test, preds, alpha=0.6, color='#2a78d6', edgecolor='white', s=50)

# diagonal reference line (perfect prediction)
min_val = min(y_test.min(), preds.min())
max_val = max(y_test.max(), preds.max())
plt.plot([min_val, max_val], [min_val, max_val], color='gray', linestyle='--', linewidth=1.5, label='Perfect prediction')

plt.xlabel('Actual Price')
plt.ylabel('Predicted Price')
plt.title('Actual vs Predicted House Price')
plt.legend()
plt.tight_layout()
plt.savefig('actual_vs_predicted.png', dpi=150)
plt.close()

# --- 4. Plot 2: Feature coefficients bar chart ---
coefs = pd.Series(model.coef_, index=X.columns).sort_values()

plt.figure(figsize=(8, 6))
colors = ['#e34948' if v < 0 else '#2a78d6' for v in coefs.values]
plt.barh(coefs.index, coefs.values, color=colors)
plt.xlabel('Effect on Price ($)')
plt.title('Feature Coefficients — Impact on House Price')
plt.axvline(0, color='black', linewidth=0.8)
plt.tight_layout()
plt.savefig('feature_coefficients.png', dpi=150)
plt.close()

print("Saved: actual_vs_predicted.png, feature_coefficients.png")
print("R2:", round(model.score(X_test, y_test), 4))
```
