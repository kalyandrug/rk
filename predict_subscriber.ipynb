```python
file_path_in_drive = '/content/drive/MyDrive/Colab Notebooks/subscriber_churn_data.csv' # Corrected path

try:
    df = pd.read_csv(file_path_in_drive)
    print('File loaded successfully!')
    display(df.head())
except FileNotFoundError:
    print(f"Error: The file '{file_path_in_drive}' was not found. Please check the path.")
except Exception as e:
    print(f"An error occurred while loading the file: {e}")
```

```python
# Basic checks
print(df.shape)
print(df.info())
print(df.isnull().sum())
```

```python
# Churn rate overall
print(df['churned_in_2_months'].value_counts(normalize=True))
```

```python
# Churn rate by category
print(df.groupby('contract_type')['churned_in_2_months'].mean())
print(df.groupby('plan_type')['churned_in_2_months'].mean())
```

```python
# Compare numeric features by churn status
numeric_cols = ['tenure_months', 'monthly_charge', 'logins_last_30d',
                 'days_since_last_login', 'support_tickets_last_90d',
                 'satisfaction_score']
print(df.groupby('churned_in_2_months')[numeric_cols].mean())
```

```python
#correlation heat map
import matplotlib.pyplot as plt
import seaborn as sns

numeric_cols = ['age', 'monthly_charge', 'tenure_months', 'autopay_enabled',
                 'discount_active', 'logins_last_30d', 'logins_prev_30_60d',
                 'days_since_last_login', 'support_tickets_last_90d',
                 'late_payments_last_12m', 'downgrade_last_6m',
                 'satisfaction_score', 'churned_in_2_months']

corr = df[numeric_cols].corr()
plt.figure(figsize=(11, 9))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('Correlation Heatmap')
plt.show()

print(corr['churned_in_2_months'].sort_values(ascending=False))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 6))
sns.boxplot(x='churned_in_2_months', y='satisfaction_score', data=df)
plt.title('Satisfaction Score Distribution by Churn Status')
plt.xlabel('Churned in 2 Months (0 = No, 1 = Yes)')
plt.ylabel('Satisfaction Score')
plt.xticks([0, 1], ['Not Churned', 'Churned'])
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
print(df.groupby('region')['churned_in_2_months'].mean() * 100)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.scatterplot(x='late_payments_last_12m', y='churned_in_2_months', data=df, alpha=0.5)
plt.title('Relationship between Late Payments and Churn')
plt.xlabel('Number of Late Payments in Last 12 Months')
plt.ylabel('Churn Status')
plt.yticks([0, 1], ['Not Churned', 'Churned']) # More descriptive labels
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.countplot(x='region', hue='churned_in_2_months', data=df, palette='viridis')
plt.title('Churn Distribution by Region')
plt.xlabel('Region')
plt.ylabel('Number of Subscribers')
plt.xticks(rotation=45) # Rotate region labels if they overlap
plt.legend(title='Churned in 2 Months', labels=['No Churn', 'Churn'])
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
